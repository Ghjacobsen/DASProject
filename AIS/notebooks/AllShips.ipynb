```python
import pandas as pd
import zipfile
# --- 1. Define Search Criteria ---
# Bounding box for the location search

#Det her er de faktiske værdier for storebælt.
min_lat = 55.33322799623468
max_lat = 55.349027891361644
min_lon = 10.976633155848155
max_lon = 11.093097420068263
zip_filename = "aisdk-2025-06-29.zip"
csv_filename = "aisdk-2025-06-29.csv" # The name of the CSV file inside the zip
# --- 2. Prepare to Process ---
chunk_size = 100000  # Process 100,000 rows at a time
found_ships = []
# --- 3. Stream and Filter ---
with zipfile.ZipFile(zip_filename) as zf:
    with zf.open(csv_filename) as f:
        # Create a text reader for pandas to process the file in chunks
        reader = pd.read_csv(f, chunksize=chunk_size, low_memory=False)
        print("Starting to process the file in chunks...")
        for i, chunk in enumerate(reader):
            print(f"Processing chunk {i+1}...")
            # Ensure column names are correct and handle potential errors
            try:
                # Filter by location (bounding box)
                # This now uses the column names 'Latitude' and 'Longitude'
                location_filtered_chunk = chunk[
                    (chunk['Latitude'] >= min_lat) &
                    (chunk['Latitude'] <= max_lat) &
                    (chunk['Longitude'] >= min_lon) &
                    (chunk['Longitude'] <= max_lon)
                ]
                if not location_filtered_chunk.empty:
                    print(f"  Found {len(location_filtered_chunk)} matching ships in this chunk!")
                    found_ships.append(location_filtered_chunk)
            except KeyError as e:
                print(f"  Warning: Column not found - {e}. Please check your CSV column names.")
                print(f"  Available columns in this chunk: {chunk.columns.tolist()}")
                break # Stop processing if columns are wrong
# --- 4. Combine and Display Results ---
if found_ships:
    final_df = pd.concat(found_ships, ignore_index=True)
    out_filename = "found_ships_june.csv"
    final_df.to_csv(out_filename, index=False)
    print("\n--- Processing Complete ---")
    print(f"Total ships found: {len(final_df)}")
    print(f"Saved results to {out_filename}")
else:
    print("\n--- Processing Complete ---")
    print("No ships were found matching your criteria.")
```
